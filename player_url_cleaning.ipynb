```python
import pandas as pd
import json
```

```python
with open ("t20_scraped_data/player_url.json") as f:
    data = json.load(f)

df_player_url = pd.DataFrame(data)

df_player_url
```

```python
df_player_url.drop_duplicates(subset='url',inplace=True)
df_player_url
```

```python
df_player_url.to_json('t20_scraped_data/player_url_cleaned.json',orient="records", force_ascii=False, indent=4)
```
